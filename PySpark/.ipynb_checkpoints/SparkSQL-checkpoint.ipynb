```python
df = spark.read.format("csv").option("header", "true").load("hdfs://localhost:9000/user/MovieReview/rotten_tomatoes_movie_reviews.csv")
```

```python
df.createTempView("df")
```

```python
spark.sql("SELECT * FROM df LIMIT 5").show()
```

```python
spark.sql("SELECT * FROM df WHERE try_cast(originalScore AS DOUBLE) <= 0.2").show()
```

```python
spark.sql("SELECT * FROM df WHERE try_cast(originalScore AS DOUBLE) > 0.5").show()
```

```python
spark.sql("SELECT COUNT(*) AS Total_Fresh FROM df WHERE reviewState == 'fresh'").show()
```

```python
spark.sql("SELECT COUNT(*) AS Total_Rotten FROM df WHERE reviewState == 'rotten'").show()
```

```python
 spark.sql("SELECT COUNT(*) AS Total_TopCritic FROM df WHERE isTopCritic == 'True'").show()
```

```python
 spark.sql("SELECT COUNT(*) AS Total_TopCritic FROM df WHERE isTopCritic == 'False'").show()
```

```python
spark.sql("SELECT DISTINCT id, creationDate FROM df").show()
```
